```python
from sklearn.datasets import fetch_covtype
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

covtype = fetch_covtype()
```

```python
print(covtype.DESCR)
```

```python
X_train, X_test, y_train, y_test = train_test_split(covtype.data, covtype.target, random_state=42)

X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```python
mlp_clf = MLPClassifier(hidden_layer_sizes=[200, 100, 50], early_stopping=True,
                        verbose=True, random_state=42)

pipeline = make_pipeline(StandardScaler(), mlp_clf)
pipeline.fit(X_train, y_train)
```

```python
sorted(MLPClassifier().get_params().keys())
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "mlpclassifier__hidden_layer_sizes" : [(100), (200, 100, (200, 100, 50))],
    "mlpclassifier__alpha" : [0.0001, 0.001, 0.01],
    "mlpclassifier__activation" : ["relu", "tanh"],
    "mlpclassifier__learning_rate" : ["constant", "invscaling", "adaptive"],
    "mlpclassifier__learning_rate_init" : [0.001, 0.01]
}

grid_searh = GridSearchCV(
    pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid_searh.fit(X_train, y_train)

print("Best parameters", grid_searh.best_params_)
print("Bästa CV-score", grid_searh.best_score_)
```
