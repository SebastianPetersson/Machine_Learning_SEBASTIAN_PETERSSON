## Machine learning landscape

***🧠 Learning paradigms***

    Supervised learning (övervakad inlärning)
Modellen tränas på data där både input och korrekt output finns.
- Self‑supervised learning — modellen skapar egna labels genom att dölja delar av input och försöka förutsäga dem (t.ex. maskera en del av en bild och låta modellen gissa vad som saknas).
- Semi‑supervised learning — kombinerar en liten mängd märkt data med en stor mängd omärkt data.

Unsupervised learning (oövervakad inlärning)

Modellen får endast input och försöker hitta mönster, strukturer eller grupperingar.

Reinforcement learning (förstärkningsinlärning)

En agent lär sig genom att interagera med en miljö och få belöningar eller straff, med målet att maximera långsiktig reward.

***🎯 Task‑typer***

- **Regression** — förutsäga kontinuerliga värden.
- **Classification** — förutsäga kategorier.
- **Clustering** — gruppera data utan labels.
- **Dimensionality reduction (komprimering)** — minska antalet features men behåll viktig information.
- **Anomaly detection** — upptäcka avvikande beteenden som inte borde hända.
- **Novelty detection** — upptäcka nya mönster som vi vill hitta.

***⚙️ Training‑strategier***

Både batch och online learning sker i praktiken offline — skillnaden ligger i hur modellen uppdateras.

- **Batch learning** — modellen tränas på hela datasetet eller stora batchar åt gången; kräver retraining vid ny data.
- **Online learning** — modellen uppdateras kontinuerligt när ny data kommer in.

***🧩 Model types***
- **Instance‑based (non‑generalizing)** — modellen lagrar instanser och jämför ny data direkt mot tidigare exempel.

Exempel: k‑Nearest Neighbors (kNN).
- **Model‑based** — modellen lär en generell funktion f(x) som kan generalisera till ny data.

Exempel: linjär regression, neurala nätverk, decision trees.

## Sammanfattning av GLM‑intuitionen

En GLM kräver tre komponenter:

- **En fördelning för \(Y\)** (normal, gamma, Bernoulli, Poisson, binomial).  
- **En linjär prediktor** $\eta = X\beta$.  
- **En länkfunktion** \(g\) som omvandlar $\mu = E(Y)$ till något som kan modelleras linjärt.

Grundidén är att välja en transformation av data så att den linjära modellen fungerar.

---

### Kontinuerliga värden över hela reala linjen
- Dataområde: \((-$\infty, +\infty$)\).  
- Fördelning: normal.  
- Länk: identitet.  
- Detta är vanlig linjär regression.

---

### Positiva kontinuerliga värden
- Dataområde: \((0, +$\infty$)\).  
- Fördelning: gamma eller exponentiell.  
- Länk: inverse eller log.  
- Användning: när variansen ökar med medelvärdet, t.ex. väntetider.  
- Kan vara användbart i anomaly detection när små förändringar ger stora relativa effekter.

---

### Binära utfall (0/1)
- Fördelning: Bernoulli.  
- Länk: logit.  
- Användning: klassificering och sannolikhetsmodellering.

---

### Antal träffar av \(n\) försök
- Fördelning: binomial.  
- Länk: logit.  
- Användning: räkna förekomster inom ett fast antal försök.

---

### Antal händelser (0, 1, 2, …)
- Fördelning: Poisson.  
- Länk: log.  
- Användning: händelser per tidsenhet, fel, kundflöden, anomaly detection.

---

## Estimering i GLM
När vi lämnar linjär regression fungerar inte minsta kvadrat längre.  
I stället används:

- **Maximum Likelihood Estimation (MLE)** som standard.  
- **Bayesianska metoder** (t.ex. Gibbs sampling, Monte Carlo) vid mer komplexa modeller eller när man vill ha posteriorfördelningar.

| Modell               | Fördelning | Dataområde            | Länkfunktion |
| -------------------- | ---------- | --------------------- | ------------ |
| Linjär regression    | Normal     | ( (-$\infty$, $\infty$) ) | Identitet    |
| Logistisk regression | Bernoulli  | {0,1}                 | Logit        |
| Binomial regression  | Binomial   | 0…n                   | Logit        |
| Poisson regression   | Poisson    | 0,1,2…                | Log          |
| Gamma regression     | Gamma      | ( (0,$\infty$) )        | Invers / Log |
