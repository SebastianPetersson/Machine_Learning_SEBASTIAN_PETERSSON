## Konvolutionella neurala nätverk (Convolutional Neural Networks, CNN)

Ett CNN använder en kärna, eller ett filter, som vandrar över en bild och utför konvolutioner på små bildområden, till exempel 4x4 pixlar. Resultatet används för att skapa indata till nästa lager. Idén är inspirerad av hur synsystemet fungerar, även om modellen i praktiken är matematiskt och tekniskt annorlunda. Hur långt filtret flyttas i varje steg kallas **stride**.

Mellan konvolutionslagren används ofta **poolinglager**. De minskar bildens rumsliga dimensioner och hjälper modellen att fokusera på de viktigaste dragen, samtidigt som beräkningarna blir effektivare.

Varje konvolutionslager innehåller vanligtvis flera filter, och varje filter skapar en egen **feature map**. Det gör att informationen snabbt växer i djup. Om ett lager använder 10 filter blir resultatet alltså 10 feature maps i nästa lager. I många CNN-arkitekturer används pooling för att minska storleken på dessa representationer innan de skickas vidare.
