### Statistics

- `$R^2$` : coefficient of determination (förklaringsgrad)
- `$\alpha$` : significance level
- `SSR` : sum of squares due to regression
- `$\chi^2$` : chi-squared

### Numerical / Error measures

- `SSE` : sum of squared errors
- `MSE` : mean squared error

$MSE = \mathrm{Bias}^2 + \mathrm{Var} + \mathrm{Brus}$

- Om variansen är för hög finns risk för overfitting.  


$MSE=$ $C(\boldsymbol{\theta}) = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$

---

Ridge regression aka **Thikhonov regularization**  

$Ridge regression=$ $MSE + \frac{\alpha}{m}\sum_{j=1}^{m} \theta_{j}^{2}$

**Bias increases** while **variance decreases**  

We expect this regression to generalize better, since we overfit less to the training data.

$\alpha$ - hyperparameter that indicates 'strength' of regularization.

Denna metod är känslig för kolinearitet. Ska man använda strikt ridge så ska man se till att kolinearitet inte är för hög. Den kommer tvinga koefficienter att spridas över alla värden, även de små.

### Elasticnet regression

$C(\theta) = MSE + r(2\alpha \sum_{i=1}^{m} |\theta_{i}| + (1-r) \frac{\alpha}{m}\sum_{j=1}^{m} \theta_{j}^{2})$

r - hyperparameter that decides how much of $l1$ (Lasso) or $l2$ (Ridge) regularization we use.

L1-norm = Lasso *(feature selection)*  l1-ratio = 1  
L2-norm= Ridge *(generalization)* l1-ratio = 0

Since we don't have (good) statistics, we must use train/test splits.

### Logistic regression

`C`: logistic loss function

- Lägger automatiskt till `L2`-regularisering (`ridge`) om man väljer det.

$\displaystyle C(\theta) = -\frac{1}{m}\sum_{i=1}^{m}\left[y_i \ln(\hat{p}_i) + (1-y_i)\ln(1-\hat{p}_i)\right] + \frac{\alpha}{2m}\sum_{j=1}^{n}\theta_j^2$  
$\displaystyle \hat{p}(x_i) = \frac{1}{1 + e^{-\theta^T x_i}}$

### Hyperparameter optimization

Thus far, we have optimized 'k' in k-NN. (i elbow plot) 

To avoid data leakage we need a validation set.  

<img src='../Data/Regularisering/Reg_2.png' width=400>

To avoid random skews (konstiga mängder), use **k-fold cross validation**.

Vi tar träningsmängden 

K-fold CV splits training data in k non overlapping subsets, called folds. (ej shufflad, endast shufflad i train-test split)

För varje fold får vi ett score

and we report the **average score**. 

Viktigt att kolla på average score, men någon av scoren kommer vara bäst, och ibland väljs den faktiskt ut, men average score ska man kolla på?
Man använder Grid search för att prova hyperparametrar, det är inget man skriver in helt manuellt för varje testkörning, utan man skriver exempelvis in en lista med olika värden för en viss hyperparameter eller flera, och skickar in i sin grid search som gör jobbet automatiskt.
