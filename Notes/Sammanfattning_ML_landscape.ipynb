## Machine Learning Landscape

En sammanfattning av centrala ML-begrepp: lärandeparadigm, modellfamiljer, optimering och utvärderingsmått.

### Learning paradigms

**Supervised learning (overvakad inlarning)**
- Modellen tränas på data där både input och korrekt output finns.
- Typiska uppgifter: regression och klassificering.

**Self-supervised learning**
- Modellen skapar egna labels från omärkt data, t.ex. genom att maskera delar av input och förutsäga det som saknas.

**Semi-supervised learning**
- Kombinerar en liten mängd märkt data med en stor mängd omärkt data.

**Unsupervised learning (oövervakad inlärning)**
- Modellen får endast input och söker mönster, struktur eller grupperingar.

**Reinforcement learning (förstärkningsinlärning)**
- En agent interagerar med en miljö, får belöningar/straff och lär en policy som maximerar långsiktig reward.

### Task-typer

- **Regression**: förutsäga kontinuerliga värden.
- **Classification**: förutsäga kategorier/klasser.
- **Clustering**: gruppera data utan labels.
- **Dimensionality reduction (komprimering)**: minska antal features men behålla viktig information.
- **Anomaly detection**: upptäcka avvikande beteenden.
- **Novelty detection**: upptäcka nya mönster som inte fanns i träningsdata.

### Training-strategier

Både batch och online kan användas i praktiken; skillnaden är hur uppdateringar görs:
- **Batch learning**: tränar på hela datasetet eller stora batchar, och behöver ofta retraining vid ny data.
- **Online learning**: uppdaterar modellen inkrementellt med ny data i små steg.

### ML och statistik

Vi gör ofta **prediktiv validering** i ML (t.ex. train/validation/test, cross-validation), medan klassisk statistik ofta fokuserar på inferens, osäkerhet och hypotesprövning.

Detta leder ibland till begreppsförvirring mellan ML och statistik. Båda behövs:
- Statistik hjälper oss förstå osäkerhet och antaganden.
- ML prioriterar ofta generalisering och prediktionsprestanda.

Tolkning av konfidensintervall: se `L1.Monte_Carlo.ipynb`.

## GLM-bakgrund: statistik till ML

**Allman linjar regression (General Linear Models, GLM i klassisk statistik)**

Klassiska verktyg:
- F-test (ANOVA, ANCOVA)
- t-test

I statistiken börjar vi ofta med en responsvariabel $Y$, men med kategoriska effekter kan modellen uttryckas med flera designkolumner och kontraster.

Exempel på regressionsmodeller som ofta presenteras i samma familj:
- linjär
- multipel
- polynomiell
- Poisson
- logistisk (utfall i [0,1] efter länkfunktion)

En klassisk linjär modell:

$Y = X\boldsymbol{\beta} + \varepsilon$

Här görs ofta normalantaganden (särskilt om feltermen $\varepsilon$).

**När vi går mot modern ML blir fokus mer på prediktion och färre starka antaganden om normalitet.**

## Generalized Linear Models (GLM)

I GLM låter vi responsen följa en fördelning ur exponential family och kopplar medelvärdet till en linjär prediktor via en länkfunktion.

<img src="../Data/gen_lin_models.png" width="400">

## Sammanfattning av GLM-intuitionen

En GLM kräver tre komponenter:
- **En fördelning för $Y$** (t.ex. normal, gamma, Bernoulli, Poisson, binomial).
- **En linjär prediktor**: $\eta = X\beta$.
- **En länkfunktion** $g$ som kopplar $\mu = E(Y)$ till den linjära prediktorn: $g(\mu)=\eta$.

Grundidén: välj fördelning + länk så att problemet blir linjärt i parametrarna.

### Vanliga fall

**Kontinuerliga värden over hela reella linjen**
- Dataområde: $(-\infty, +\infty)$
- Fördelning: normal
- Länk: identitet
- Exempel: vanlig linjär regression

**Positiva kontinuerliga värden**
- Dataområde: $(0, +\infty)$
- Fördelning: gamma/exponentiell-liknande
- Länk: inverse eller log
- Exempel: väntetider, kostnader

**Binära utfall (0/1)**
- Fördelning: Bernoulli
- Länk: logit
- Exempel: klassificering

**Antal träffar av $n$ försök**
- Fördelning: binomial
- Länk: logit

**Antal händelser (0,1,2,...)**
- Fördelning: Poisson
- Länk: log
- Exempel: antal fel per tidsenhet, antal kunder per tidsfönster

När vanlig minsta kvadrat inte är lämplig används ofta **Maximum Likelihood Estimation (MLE)**, numeriskt med iterativa metoder.

### Gradient Descent (se `L2.Gradient_descent.ipynb`)

Analytiskt försöker vi ofta derivera och sätta derivatan till 0. I många ML-problem gör vi i stället numerisk optimering.

**Gradient Descent (GD)**
- Uppdaterar parametrar iterativt i riktning mot minskad kostnad.
- Kräver en **learning rate** ($\eta$), dvs steglängd.

Om learning rate är:
- för liten -> långsam konvergens
- för stor -> risk att hoppa över optimum eller divergera

GD behöver stoppvillkor, t.ex.:
- max antal iterationer
- liten förbättring under flera steg
- liten gradientnorm

<img src="../Data/Learningrate1.png" width="600">

Flera optimum kan förekomma: **lokala** och **globala** minima.

### ADAM - Adaptive Moment Estimation

ADAM kombinerar:
- momentum ("tröghet" från tidigare gradienter)
- adaptiva steglängder per parameter

Det gör optimeringen stabilare i många praktiska problem.

<img src="../Data/ADAM.png" width="500">

I flera dimensioner använder vi **gradienten** i stället för enkel derivata.

"Gradient points in the direction of steepest increase."
Vid minimering går vi i motsatt riktning.

<img src="../Data/3D.png" width="200">

Om vi använder hela datasetet i varje uppdatering får vi:

**Batch Gradient Descent**
- beräknar gradient med alla datapunkter
- stabil men kan vara dyr för stora dataset

En standarduppdatering:

$\theta_{n+1} = \theta_n - \eta \, \nabla_{\theta}\, J(\theta)$

(där $J(\theta)$ kan vara t.ex. MSE eller log-loss)

**Stochastic Gradient Descent (SGD)**
- använder en slumpad datapunkt (eller mycket liten batch) per steg

<img src="../Data/StochasticGD.png" width="200">

SGD kan vara brusig eftersom uppdateringen baseras på slumpade observationer, men är ofta snabb och kan hjälpa modellen att ta sig ur grunda lokala minima/sadelpunkter.

**Mini-batch Gradient Descent**

- använder en slumpad liten batch av datapunkter per uppdatering
- kombinerar stabiliteten från batch med hastigheten från SGD

Detta är den vanligaste varianten i praktisk träning av ML-modeller.

För regression utvärderar vi ofta:
- **MAE** (Mean Absolute Error)
- **MSE** (Mean Squared Error)
- **RMSE** (Root Mean Squared Error)

Kort intuition:
- MAE är mer robust mot outliers
- MSE/RMSE straffar stora fel hårdare

### Skillnad mellan loss function och cost function

Prediktion (enkel modell):

$\hat{y} = wx$

Fel för en observation:

$(\hat{y} - y)^2$

**Loss** för observation $i$:

$\ell_i(w) = (w x_i - y_i)^2$

**Cost/Objective** för hela datasetet (t.ex. medelvärde):

$J(w)=\frac{1}{n}\sum_{i=1}^{n}\ell_i(w)$

Gradientuppdatering använder learning rate $\eta$:

$w \leftarrow w - \eta\,\nabla J(w)$

## Precision and Recall

Recall:

$\frac{TP}{TP + FN}$

"Av alla riktiga bränder, hur många upptäckte vi?"

Precision:

$\frac{TP}{TP + FP}$

"Hur ofta var det faktiskt brand när larmet gick?"

Accuracy:

$\frac{TP + TN}{TP + TN + FP + FN}$

"Hur ofta hade modellen rätt totalt?"

### F1-score

$\displaystyle F_1 = \frac{2}{\frac{1}{\text{Precision}} + \frac{1}{\text{Recall}}}$

$\displaystyle F_1 = \frac{2 \cdot \text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$

F1 är användbart när klassfördelningen är obalanserad och vi vill balansera precision och recall.

### Precision/Recall Trade-Off

När beslutströskeln ändras förändras precision och recall i motsatt riktning.
- högre tröskel -> ofta högre precision, lägre recall
- lägre tröskel -> ofta högre recall, lägre precision

Välj tröskel efter verksamhetskostnad för FP respektive FN.
