# Klassificering

Klassificering handlar om att prediktera en **diskret klass** (etikett), till skillnad från regression som predikterar ett kontinuerligt värde.

Exempel:
- Spam / inte spam
- Siffra 0-9 i handskriven bild
- Sjuk / frisk

En vanlig modell är logistisk regression:

$P(y=1\mid x)=\sigma(x^T\beta)=\dfrac{1}{1+e^{-x^T\beta}}$

Beslutsregel vid tröskel $t$:

$\hat y = 1$ om $P(y=1\mid x)\ge t$, annars $\hat y=0$.

## Konfusionsmatris

För binär klassificering sammanfattar konfusionsmatrisen:
- **TP**: korrekt positiv
- **TN**: korrekt negativ
- **FP**: falskt positiv
- **FN**: falskt negativ

<img src="../Data/Klassificering/Klassificering_2.png" width="700">

## Viktiga mått

Accuracy:

$\text{Accuracy}=\dfrac{TP+TN}{TP+TN+FP+FN}$

Precision:

$\text{Precision}=\dfrac{TP}{TP+FP}$

Recall (Sensitivity, TPR):

$\text{Recall}=\dfrac{TP}{TP+FN}$

Specificity (TNR):

$\text{Specificity}=\dfrac{TN}{TN+FP}$

F1-score:

$F_1 = \dfrac{2\cdot \text{Precision}\cdot \text{Recall}}{\text{Precision}+\text{Recall}}$

Kommentar: Vid obalanserade klasser är precision/recall/F1 ofta mer informativa än accuracy.

## Tröskel och trade-off

Modellen ger ofta en score eller sannolikhet. När vi ändrar tröskeln $t$ får vi en trade-off:
- Högre tröskel: färre positiva prediktioner, ofta högre precision men lägre recall.
- Lägre tröskel: fler positiva prediktioner, ofta högre recall men lägre precision.

Val av tröskel ska styras av kostnaden för FP respektive FN i problemet.

## ROC- och PR-perspektiv

- **ROC-kurva**: TPR mot FPR, där $\text{FPR}=\dfrac{FP}{FP+TN}$.
- **ROC AUC**: sannolikheten att modellen rankar ett positivt exempel högre än ett negativt.
- **Precision-Recall-kurva**: särskilt användbar vid stark klassobalans.

Praktisk tumregel:
- Balanserade data: ROC AUC fungerar ofta bra.
- Obalanserade data: PR-kurvan och F1 är oftast mer relevanta.

## Multiklassklassificering

Två standardstrategier:
- **One-vs-Rest (OvR / OvA)**: en binär klassificerare per klass mot resten.
- **One-vs-One (OvO)**: en klassificerare per klasspar.

Softmax-modeller (multinomial logistisk regression) predikterar sannolikhetsfördelning över alla klasser direkt.

<img src="../Data/Klassificering/Klassificering_5.png" width="600">

## Generativa klassificerare

### LDA och QDA
- **LDA** antar normalfördelade klasser med gemensam kovariansmatris.
- **QDA** antar normalfördelade klasser men separat kovarians per klass.
- LDA ger linjära beslutsgränser, QDA kvadratiska.

### Naive Bayes
Naive Bayes bygger på Bayes sats och antagandet att features är villkorligt oberoende givet klass:

$P(Y=k\mid X=x)=\dfrac{P(X=x\mid Y=k)P(Y=k)}{P(X=x)}$

Vanliga varianter: Gaussian, Multinomial och Bernoulli Naive Bayes.

## K-NN (k-nearest neighbors)

K-NN är en icke-parametrisk, instansbaserad metod:
- Hitta de $k$ närmaste träningspunkterna.
- Låt dessa rösta (majoritetsröstning) eller vikta med avstånd.

Egenskaper:
- Känslig för featureskalning: standardisering är ofta nödvändig.
- Litet $k$: låg bias, hög varians.
- Stort $k$: högre bias, lägre varians.

<img src="../Data/Klassificering/Klassificering_6.png" width="300">

K-NN används bland annat för bildklassificering (t.ex. handskrivna siffror), rekommendation och sökproblem i feature-rum.

## Avståndsmått och likhet

För många klassificerare behöver vi ett avstånd/likhetsmått mellan observationer.

Manhattan distance (L1):

$d_1(a,b)=\sum_i |a_i-b_i|$

Euclidean distance (L2):

$d_2(a,b)=\sqrt{\sum_i (a_i-b_i)^2}$

Val av mått påverkar både prestanda och robusthet.

<img src="../Data/Klassificering/Klassificering_7.png" width="220">

L1 och L2 ger olika geometri i feature-rummet, vilket i sin tur påverkar närmaste grannar och beslutsgränser.

## Cosine similarity och textklassificering

I högdimensionell textdata är vinkel ofta viktigare än längd.

Cosine similarity:

$\text{cosine}(u,v)=\dfrac{u\cdot v}{\|u\|\,\|v\|}=\cos\alpha$

Därför används ofta:
- **TF-IDF** för klassisk textrepresentation.
- **Embeddings** (t.ex. från transformerbaserade modeller) för modern representation.

## Modellval och Occams rakkniv

Occams rakkniv: om flera modeller presterar likvärdigt, välj den enklare modellen.

Praktiskt innebär det ofta:
- färre antaganden
- lägre komplexitet
- bättre tolkbarhet
- mindre risk för överanpassning

## Logistisk regression vs K-NN

Ingen metod "vinner alltid"; valet beror på datans struktur.

**Logistisk regression** passar ofta bättre när:
- sambandet mellan features och klass är ungefär linjärt i beslutsytan
- du vill ha snabb inferens och god tolkbarhet
- du vill extrapolera mer stabilt utanför täta kluster

**K-NN** passar ofta bättre när:
- beslutsgränserna är tydligt icke-linjära och lokala
- liknande observationer verkligen ligger nära varandra i feature-rummet
- du har tillräckligt med representativa träningspunkter

Obs: K-NN blir ofta dyrt vid prediktion på stora dataset eftersom avstånd till många punkter måste beräknas.

## Val av $k$ och skalning i K-NN

K-NN har hyperparametern $k$ (antal grannar), men ofta väljer man även:
- avståndsmått (t.ex. L1/L2)
- viktning (uniform eller avståndsbaserad)

Val av $k$ görs lämpligen med korsvalidering och valideringskurva ("elbow"/score-kurva):
- litet $k$ -> känslig modell (hög varians)
- stort $k$ -> slätare modell (högre bias)

Feature-skalning är central för K-NN eftersom avstånd domineras av variabler med stor skala.
Använd i praktiken oftast **StandardScaler** eller **MinMaxScaler** beroende på data.
