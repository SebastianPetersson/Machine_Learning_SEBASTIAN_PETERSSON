## Similarity

ML-systems need numerical data, so to be able to measure similarity:

$Y_i = g^{-1}(X\beta)$  
Similar y's must have similar $X_i\beta_i$  

If so, then the following measures **distance**  |a + b| = $l$ -norm or taxi-cab (manhattan distance)  
(x,y) = |x + y|

$\sqrt{a^2 + b^2}$ = $l_2$ - norm, euclidean distance 

<img src="../Data/Klassificering_7.png" width="100">  


rather than a normal cartesian system (normal coordinates) we can use **polar coordinates**  
*(note: vectors are position-independent)*  
<img src="../Data/Klassificering_8.png" width="300">  

only the angle is now relevant!

$\dfrac{u \cdot v}{\|u\|\,\|v\|} = \cos\alpha$ = **cosinus similarity**

Text is favourably compared using cosine-similarity, but we need **vectorization**

**Classically**: TF - IDF  

**Term Frequency / Inverse Document Frequency** *(en teknik från lingvistik)*

**Modern**: token embeddings in LLMs

Most of these techniques are instance-based (ie don't generalize well)  
(de är känsliga för förändringar, outliers osv)  
eq. If you trained on a corpus of technical manuals, you won't be able to classify fantasy books.


**Occams rakkniv** (Ockhams rakkniv) är en filosofisk och vetenskaplig princip som förespråkar enkelhet vid val av förklaringar. Den innebär i korthet att om det finns flera möjliga förklaringar till ett fenomen, bör man välja den som kräver færrest antaganden.
