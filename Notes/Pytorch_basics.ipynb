## Vad är neurala nätverk?

Neurala nätverk är modeller inom maskininlärning som inspirerades av en tidig, förenklad idé om hur nervceller fungerar. De utvecklades ursprungligen utifrån sådana idéer redan på 1950-talet.

Ett neuralt nätverk består av noder, som ofta kallas **neuroner**, organiserade i olika lager. Varje koppling mellan noder har en **vikt**, som bestämmer hur starkt en signal påverkar nästa lager.

För en nod gör man först en linjär beräkning:

$z = XW + b$

där:
- $X$ är indata
- $W$ är vikterna
- $b$ är bias

Sedan skickas resultatet genom en **aktiveringsfunktion**:

$\hat{y} = f(XW + b)$

Aktiveringsfunktionen gör modellen icke-linjär, vilket är viktigt eftersom nätverket annars bara hade fungerat som en vanlig linjär modell.

## MLP

**MLP** står för **Multilayer Perceptron**. Det är ett vanligt framåtkopplat neuralt nätverk där informationen går från indata till utdata genom ett eller flera dolda lager.

## Begrepp

**Aktiveringsfunktion**  
En funktion som appliceras efter den linjära beräkningen. Vanliga exempel är ReLU, sigmoid och tanh. Den hjälper nätverket att lära sig mer komplexa samband.

**Rekurrenta lager**  
Rekurrenta lager används när ordningen i datan spelar roll, till exempel i text, tidsserier eller tal. De kan ta hänsyn till tidigare information i sekvensen. Denna typ av nätverk kallas ofta **RNN**.

**Skip connections**  
Skip connections innebär att information får hoppa över ett eller flera lager i nätverket. Det hjälper modellen att bevara viktig information och gör det lättare att träna djupa nätverk. Detta används bland annat i ResNet.

ReLU står för `Rectified Linear Unit` och är en vanlig aktiveringsfunktion i neurala nätverk.

Definition:
`$\mathrm{ReLU}(x) = \max(0, x)$`

Det betyder:
- om `x < 0` blir output `0`
- om `x > 0` blir output `x`

Exempel:
- `ReLU(-3) = 0`
- `ReLU(2) = 2`

Varför den används:
- den är enkel och snabb att beräkna
- den fungerar ofta bra i djupa neurala nätverk
- den hjälper modellen att lära sig icke-linjära samband

Vilka vikter modellen får från början är mycket viktigt. Små skillnader i startvärden kan leda till att träningen utvecklas på olika sätt och att modellen till slut hamnar i olika lösningar.

Därför använder man ofta genomtänkta metoder för viktinitialisering, till exempel **Xavier-initialisering** eller **He-initialisering**. De hjälper modellen att få en bättre start och kan göra träningen mer stabil.

Under träningen jämförs modellens prediktion $\hat{y}$ med det sanna värdet $y$. Skillnaden mellan dessa används för att beräkna felet.

För all träningsdata uppdateras vikterna sedan iterativt med gradient descent:

$w \leftarrow w - \eta \nabla C(w)$

där:
- $w$ är vikterna
- $\eta$ är inlärningshastigheten
- $\nabla C(w)$ är gradienten av kostnadsfunktionen

## NLP

**NLP** står för **Natural Language Processing** och handlar om att låta datorer bearbeta och förstå mänskligt språk.

Många klassiska NLP-tekniker som användes före 2017 var användbara, men de räcker ofta inte för dagens mer avancerade språkmodeller.

**GPT** står för **Generative Pre-trained Transformer**.
- **Generative** - att modellen kan generera text
- **Pre-trained** - att den först tränas på mycket stora mängder text
- **Transformer** - är den modellarkitektur som används

En transformer består i grunden av en **encoder** och en **decoder**, även om vissa modeller bara använder en del av arkitekturen. Mellan dessa sker flera beräkningar i dolda lager där modellen lär sig relationer i texten.

## Embeddings

För att en modell ska kunna arbeta med text måste orden först omvandlas till numeriska representationer.

En **embedding** är en numerisk representation av ett ord eller en token. Den fångar information om ordets betydelse och relation till andra ord.

När texten delas upp i mindre delar som modellen arbetar med kallas dessa delar för **tokens**. En token kan vara:
- ett helt ord
- en del av ett ord
- ett skiljetecken

Embeddings gör alltså att text kan representeras som tal, så att modellen kan bearbeta den matematiskt.


## Backpropagation i Torch

I PyTorch sker backpropagation med hjälp av **autograd**, som är PyTorchs system för automatisk derivering. Det använder **reverse-mode automatic differentiation**, vilket betyder att PyTorch först räknar ut resultatet i ett forward-pass och sedan arbetar baklänges genom beräkningarna för att ta fram gradienterna.

Om vi tar ett enkelt exempel:

- `f(x) = x^2`
- `f(5) = 25`
- `f'(x) = 2x`
- `f'(5) = 10`

Det betyder att om `x = 5`, så är funktionens värde `25`, och lutningen i den punkten är `10`. Gradientens värde talar alltså om hur mycket funktionsvärdet förändras om vi ändrar `x` lite grann.

I Torch markerar man att en tensor ska spåras genom att sätta `requires_grad=True`. Då bygger PyTorch upp en beräkningsgraf under forward-passet. När vi sedan anropar `.backward()` räknas derivatorna ut automatiskt, och gradienten lagras i `.grad`.

I just det här exemplet skulle Torch alltså komma fram till att gradienten är `10`. Det är samma sak som den matematiska derivatan `f'(5)`.

Detta är grunden i backpropagation när vi tränar neurala nätverk: PyTorch räknar ut hur mycket varje parameter påverkar felet, och sedan kan en optimerare uppdatera vikterna i rätt riktning.

nn.Sequential och nn.Linear är två vanliga byggstenar i PyTorch när man skapar neurala nätverk.

nn.Linear används för att definiera ett fullt kopplat lager. Det betyder att varje input kopplas till varje output genom vikter och bias.

nn.Sequential är ett enkelt sätt att bygga en modell genom att lägga flera lager efter varandra i ordning. Det passar bra när modellen består av en rak kedja av lager, till exempel flera Linear-lager med aktiveringsfunktioner emellan.

I torch.utils.data finns bland annat TensorDataset och DataLoader.

TensorDataset används för att samla ihop exempelvis indata och etiketter till ett dataset som PyTorch kan arbeta med.

DataLoader används för att läsa in datan i mindre grupper, så kallade mini-batches. Det gör träningen mer effektiv, särskilt när datasetet är stort. Med DataLoader kan man också blanda datan mellan epoker genom att använda shuffle=True.

Mini-batches innebär alltså att modellen inte tränas på hela datasetet samtidigt, utan på mindre delar i taget. Det är ofta både snabbare och mer praktiskt vid träning av neurala nätverk.
